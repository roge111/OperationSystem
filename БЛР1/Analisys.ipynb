```python
import pandas as pd
import matplotlib.pyplot as plt


```

```python
filename = int=input("Введите имя файла: ")

data = pd.read_csv(filename)

df = pd.DataFrame(data)
```

```python
plt.figure(figsize=(10, 7))
scatter = plt.scatter(df['Количество нагрузки на CPU'],
                    df['Количетсво нагрузки на память'],
                    c=df['Время работы'],
                    cmap='plasma', alpha=0.8)
plt.colorbar(scatter, label='Время работы')

plt.xlabel('Количество нагрузки на CPU')
plt.ylabel('Количетсво нагрузки на память')
plt.title('Время работы (цвет) от CPU и памяти')
plt.show()
```

```python
plt.plot(df['Количетсво нагрузки на память'], df['%USER'])
plt.xlabel('Количество нагрузки на память')
plt.ylabel('%USER')
plt.show()
```

```python
plt.plot(df['Количество нагрузки на CPU'], df['%USER'])
plt.xlabel('Количество нагрузки на CPU')
plt.ylabel('%USER')
plt.show()
```

```python
plt.plot(df['Количество нагрузки на CPU'], df['%SYS'])
plt.xlabel('Количество нагрузки на CPU')
plt.ylabel('%SYS')
plt.show()
```
