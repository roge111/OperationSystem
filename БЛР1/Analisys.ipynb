```python
import pandas as pd
import matplotlib.pyplot as plt


```

```python
filename = int=input("Введите имя файла: ")

data = pd.read_csv(filename)

df = pd.DataFrame(data)


pd.set_option('display.max_rows', None)    # показать все строки
pd.set_option('display.max_columns', None)  # показать все столбцы
pd.set_option('display.width', None)         # полная ширина вывода
pd.set_option('display.max_colwidth', None) # полная ширина столбцов (без обрезания текста)


df  # теперь вывод будет полным
```

```python
def draw_plot(name_column: str):
    plt.figure(figsize=(10, 7))
    scatter = plt.scatter(df['Количество нагрузки на CPU'],
                        df['Количетсво нагрузки на память'],
                        c=df[name_column],
                        cmap='plasma', alpha=0.8)
    plt.colorbar(scatter, label=name_column)

    plt.xlabel('Количество нагрузки на CPU')
    plt.ylabel('Количетсво нагрузки на память')
    plt.title(f'{name_column} (цвет) от CPU и памяти')
    plt.show()
```

```python

draw_plot('Время работы')
```

```python
draw_plot('%USER')
```

```python
draw_plot('%SYS')
```

```python
draw_plot('%WAIT')
```

```python
draw_plot('Количество переключений контекста')
```

```python
draw_plot('Количество паралельных процессов')
```
